```python
#box divided into two equal halves, at t = 0 there are N = 5000 particles on the left side. a hole is punched one particle can pass through at a time. all have equal probability of going to either side graphically determining equilibrium state when sufficient time has passed

#PROGRAM STARTS HERE

from library import random
import matplotlib.pyplot as plt

rng = random()
t = 0
left = 5000
right = 0
T = []
Nleft = []
Nright = []

T.append(t)
Nleft.append(left)
Nright.append(right)

while t < 5000:
    mr = 0
    for n in range(left):
        if rng.next_rand() < 0.005:
            mr = mr + 1     
    ml = 0
    for n in range(right):
        if rng.next_rand() < 0.005:
            ml = ml + 1  
    left = left - mr + ml
    right = right + mr - ml
    t = t + 1
    
    T.append(t)
    Nleft.append(left)
    Nright.append(right)

print('plot for particles in left partition in red and for right in blue')
plt.scatter(T, Nleft, color='red', s=1)
plt.scatter(T, Nright, color='blue', s=1)
plt.show()

#PROGRAM ENDS HERE
```

```python
#finding inverse of matrix using LU decomposition

#PROGRAM STARTS HERE

import library

A = []
with open('asgn0_matA','r') as f:
    ra = f.read()
    ral = list(ra.strip(' ').split('\n'))
    ral.pop()
    for i in ral:
        row0 = []
        row = i.split(' ')
        for j in row:
            try:
                row0.append(float(j))

            except:
                continue
        A.append(row0)
LU_A = library.LU(A)
n = len(A)

A_inv = []
for i in range(n):
    row = []
    for j in range(n):
        row.append(0.0)
    A_inv.append(row)

for j in range(n):
    b = []
    for i in range(n):
        if i == j:
            b.append(1.0)
        else:
            b.append(0.0)

    x = library.forback(LU_A, b)

    for i in range(n):
        A_inv[i][j] = round(x[i], 3)

print(A_inv)

#PROGRAM ENDS HERE
```

```python
#weins displacement lawequation solved with newton raphson method 
# we need to find the root of the plancks function then substitute to find the value of b

#PROGRAM STARTS HERE

import math
import library

def f(x):
    return (x - 5) * math.exp(x) + 5

root = library.bisection(f, 4, 5)

x_val = root[0]
h = 6.626e-34
k = 1.381e-23
c = 3e8  
b = (h * c) / (x_val * k)
    
print("Wien's Constant b: ", b, "m K")

#PROGRAM ENDS HERE
```

```python
# to find the centre of mass of 2 meter beam where denisty function is given as x^2
# using function for moment of inertia

# PROGRAM STARTS HERE
import library

def lamda(x):
    return x**2

def numerator(x):
    return x * lamda(x)

mass = library.simpson(lamda, 0, 2, 100)
moment = library.simpson(numerator, 0, 2, 100)

com = moment / mass

print("Centre of mass =", com)
# PROGRAM ENDS HERE
```
