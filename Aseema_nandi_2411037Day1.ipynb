```python
#Use iterative eqn xi+1 = cxi(1 - xi) to generate 1000 random numbers. Show correlation by plotting x1 vs xi+k. using x0 = 0.1 and 3 different c. and various k

#PRGOGRAM STARTS HERE

import matplotlib.pyplot as plt

x0 = 0.1 
listc = [3.22, 3.94, 4]
listk = [2, 9, 5]
random_series = []


for c in listc:
    rand = []
    xi = x0  
    for i in range(0, 100):
        xi1 = c * xi * (1 - xi)
        rand.append(xi1)
        xi = xi1
    random_series.append(rand)

for k in listk:
    for i, c in enumerate(listc):
        x_data = random_series[i][0 : 100 - k]
        y_data = random_series[i][k : 100]
        print('c = ',c,'k = ',k)
        plt.figure(figsize=(8, 5))
        plt.scatter(x_data, y_data)
        
#PROGRAM ENDS HERE
        

```

```python
#Simulate the radioactive decay A → B → C for the decay constants λA = 0.0347 and λB = 0.0198. Initial number of nuclei NA = 500, NB = 0, NC = 0.
#Uses updated rng function from library

#PROGRAM STARTS HERE

import matplotlib.pyplot as plt
from library import random
NA = 500
NB = 0
NC = 0
lamdaA = 0.0347
lamdaB = 0.0198
rng = random()
la = []
lb = []
lc = []
t = []
for i in range(0,600):
    Nac = NA
    Nbc = NB
    for j in range(0,Nac):
        k = rng.next_rand()
        if k < lamdaA:
            NA = NA-1
            NB = NB+1
    for j in range(0,Nbc):
        k = rng.next_rand()
        if k < lamdaB:
            NB = NB-1
            NC = NC+1
    la.append(NA)
    lb.append(NB)
    lc.append(NC)
    t.append(i)
plt.scatter(t,la)
plt.scatter(t,lb)
plt.scatter(t,lc)

#PROGRAM ENDS HERE
```

```python
#The updated rng code used

class random:
    def __init__(self, seed=0.1):
        self.a = 1103515245
        self.c = 12345
        self.m = 32768
        self.x0 = seed

    def next_rand(self):
        xi = ((self.a * self.x0 + self.c) % self.m) / self.m
        self.x0 = xi
        return xi
```
