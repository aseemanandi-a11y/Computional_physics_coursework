```python
#Prepare library routines for Cholesky decomposition, which must include a check for symmetric matrix.

#PROGRAM STARTS HERE

import math

def cholesky(A):
    #check if matrix is symmetric
    for i in range (len(A)):
        for j in range(len(A)):
            if A[i][j] != A[j][i]:
                return 'matrix is not symmetric'
    for i in range(len(A)):
        for j in range(i+1):
            sum = 0
            for k in range (j):
                sum = sum + (A[i][k]*A[j][k])
            if i == j:
                A[i][j] = math.sqrt(A[i][j] - sum)
            else:
                A[i][j] = (A[i][j] - sum)/(A[j][j])
                A[j][i] = A[i][j]

    return A

#PROGRAM ENDS HERE
```
