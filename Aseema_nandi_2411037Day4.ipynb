```python
#library code for jacobi

#PROGRAM STARTS HERE

def jacobi(A,b,tol = 1e-6):

    #checking diagonal dominance:
    for i in range (len(A)):
        s = 0
        for j in range(len(A)):
            if j != i:
                s = s+A[i][j]
        if A[i][i] <= s:
            print('the matrix is not diagonally dominant')
            diagonal = False
            return

    #solving the equation
    xn = [0] * len(b)
    x0 = [0] * len(b)

    for k in range(1, 501): 
        for i in range(len(b)):
            sum = 0.0
            for j in range(len(b)):
                if j != i:
                    sum = sum + A[i][j] * x0[j]

            xn[i] = (b[i] - sum) / A[i][i]

        converged = True
        for i in range(len(b)):
            if abs(xn[i] - x0[i]) >= tol:
                converged = False
                break
        if converged == True :
            return xn,k

        x0 = list(xn)
    return xn,k

#PROGRAM ENDS HERE
```

```python
#library code for gauss seigel and SOR, for gauss seigel dont run any value of w

#PROGRAM STARTS HERE

def SOR(A,b,tol = 1e-6,w = 1):

    #checking diagonal dominance:
    for i in range (len(A)):
        s = 0
        for j in range(len(A)):
            if j != i:
                s = s+A[i][j]
        if A[i][i] <= s:
            print('the matrix is not diagonally dominant')
            diagonal = False
            return

#solving the equations
    x0 = [0] * len(b)

    for k in range(1, 501): 
        xtemp = list(x0)
        for i in range(len(b)):
            sum = 0.0
            for j in range(len(b)):
                if j != i:
                    sum = sum + A[i][j] * x0[j]

            x0[i] = (1 - w) * xtemp[i] + w * ((b[i] - sum) / A[i][i])

        converged = True
        for i in range(len(b)):
            if abs(x0[i] - xtemp[i]) >= tol:
                converged = False
                break
        if converged == True :
            return x0,k

        x0 = x0
    return x0,k

#PROGRAM ENDS HERE
```
