```python
# library code forsimpson one third

#PROGRAM STARTS HERE

def simpson(f, a, b, n):
  if n % 2 != 0:
    raise ValueError("n must be an even number.")
  h = (b - a) / n
  sum = f(a) + f(b)
  for i in range(1, n):
    k = a + i * h
    if i % 2 == 0:
      sum = sum + 2 * f(k)
    else:
      sum = sum + 4 * f(k)
  return sum * h / 3

#PROGRAM ENDS HERE
```
