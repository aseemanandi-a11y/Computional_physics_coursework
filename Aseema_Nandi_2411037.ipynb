```python
#calcultae the sum of N=15 terms of GP and HP with common ration 0.5 nd common difference 1.5 starting from t0=1.25

#PRGOGRAM STARTS HERE

#geometric progression
t0 = 1.25
sum = 1.25
for i in range (0,15):
    t = t0*0.5
    sum = sum + t
    t0 = t
print(sum)

#harmonic progression
t0 = 1.25
sum = 1.25
for i in range(0,15):
    t = t0/(1.5*t0+1)
    sum = sum + t
    t0 = t
print(sum)

#PROGRAM ENDS HERE
```

```python
# given matrices A,B,C,D find AB,D.C,BC

#PRGOGRAM STARTS HERE

matA = []
matB = []
vecC = []
vecD = []
with open('asgn0_matA','r') as f:
    ra = f.read()
    ral = list(ra.strip(' ').split('\n'))
    ral.pop()
    for i in ral:
        row0 = []
        row = i.split(' ')
        for j in row:
            try:
                row0.append(float(j))

            except:
                continue
        matA.append(row0)

with open('asgn0_matB','r') as f:
    rb = f.read()
    rbl = list(rb.strip(' ').split('\n'))
    rbl.pop()
    for i in rbl:
        row0 = []
        row = i.split(' ')
        for j in row:
            try:
                row0.append(float(j))

            except:
                continue
        matB.append(row0)

with open('asgn0_vecC','r') as f:
    rc = f.read()
    rcl = list(rc.strip(' ').split('\n'))
    rcl.pop()
    for i in rcl:
        vecC.append(float(i))

with open('asgn0_vecD','r') as f:
    rd = f.read()
    rdl = list(rd.strip(' ').split('\n'))
    rdl.pop()
    for i in rdl:
        vecD.append(float(i))

#calculating AB
row1 = []
AB = []
val = 0
for i in range (0,3):
    for k in range(0,3):
        for j in range(0,3):
            val = val + matA[i][j]*matB[j][k]
        row1.append(val)
        val = 0
    AB.append(row1)
    row1 = []
print(AB)

#calculating D.C
ip = 0
for i in range(0,3):
    ip = ip + vecD[i]*vecC[i]
print(ip)

#calculating BC
BC = []
val = 0
for i in range (0,3):
    for j in range(0,3):
        val = val + matB[i][j]*vecC[j]
    BC.append(val)
    val = 0
print(BC)

#PROGRAM ENDS HERE
```

```python
#define mycomplex class/structure and calculate the sum,difference,product and modulus of (1.3-2.2i)and(-0.8+1.7i)

#PRGOGRAM STARTS HERE

import math
class MyComplex:
    def __init__(self,real,complex):
        self.real = real
        self.complex = complex

    def display(self):
        return(self.real, '+', self.complex, 'i')
    
    def addcmplx(self,c2):
        c = MyComplex(self.real+c2.real,self.complex+c2.complex)
        return(c)
    
    def subcmplx(self,c2):
        c = MyComplex(self.real-c2.real,self.complex-c2.complex)
        return(c)
    
    def prodcmplx(self,c2):
            c = MyComplex(self.real*c2.real - self.complex*c2.complex, self.real*c2.complex + self.complex*c2.real)
            return(c)

    def modulus(self):
         r = self.real**2
         c = self.complex**2
         print(math.sqrt(r+c))

a = MyComplex(1.3,-2.2)
b = MyComplex(-0.8,1.7)
print(a.addcmplx(b).display())
print(a.subcmplx(b).display())
print(a.prodcmplx(b).display())
a.modulus()
b.modulus()

#PROGRAM ENDS HERE

```
