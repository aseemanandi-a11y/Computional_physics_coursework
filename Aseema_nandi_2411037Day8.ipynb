```python
# library function for numerical integration with midpoint

#PROGRAM STARTS HERE

def midpoint(f, a, b, n = 10):
    h = (b - a) / n
    sum = 0
    for i in range(n):
        sum = sum + f(a + (i + 0.5) * h)
    return sum * h


#PROGRAM ENDS HERE
```

```python
# library function for numerical integration with trapezoidal

#PROGRAM STARTS HERE

def trapezoidal(f, a, b, n = 10):
    h = (b - a) / n
    sum = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        sum = sum + f(a + i * h)
    return sum * h

# Example usage: integrate x^2 from 0 to 2 with n = 4
def f(x):
    return x

#PROGRAM ENDS HERE
```
