```python
#write library function for Laguerre's method

#PROGRAM STARTS HERE

import math

def derivatives(coeffs,x0):
    p = 0
    dp = 0
    ddp = 0
    for c in coeffs:
        ddp = ddp * x0 + 2 * dp
        dp = dp * x0 + p
        p = p * x0 + c
        
    return p, dp, ddp
    


def laguerre_method(coeffs, x0, tol=1e-12):
    n = len(coeffs) - 1
    if n <= 0:
        return None
        
    x = x0
    
    for i in range(100):
        p,dp,ddp = derivatives(coeffs,x)
        
        if abs(p) < tol:
            return x
            
        G = dp/p
        H = G**2 - ddp/p
        
        inner = (H * n - G**2) * (n - 1)
        inner_sqrt = math.sqrt(inner)

        denom1 = G + inner_sqrt
        denom2 = G - inner_sqrt
        denom = denom1 if abs(denom1) > abs(denom2) else denom2
        
        if denom == 0.0:
            a = 0
        else:
            a = n/denom
            
        # x = x - a
        x = x - a
        
        if a < tol:
            break
            
    return x

#PROGRAM ENDS HERE
```

```python
#write library function for deflation

#PROGRAM STARTS HERE

def deflate_polynomial(coeffs, root):
    deflated = []
    remainder = 0
    for c in coeffs[:-1]:
        remainder = remainder*root + c
        deflated.append(remainder)
    return deflated

#PROGRAM ENDS HERE
```
