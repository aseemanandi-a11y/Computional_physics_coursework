```python
# library function for newton raphson

#PROGRAM STARTS HERE

def newton_raphson(f,df,x0,delta = 1e-6):

    x = x0

    #starting iterations with maximum 1000 iterations.
    for i in range(1000):
        fx = f(x)
        dfx = df(x)
        if abs(dfx) < 1e-6:
            print("Derivative too close to zero. Method fails.")
            return None
        x1 = x - (fx/dfx)
        if abs(x1 - x) < delta:
            return x1,i
        x = x1
    print('method did not converge after 1000 iterations')
    return x,1000

#PROGRAM ENDS HERE
```

```python
# library function for fixed point

#PROGRAM STARTS HERE

def fixed_point(g,x0,delta=1e-6):
    for i in range(1000):
        x1 = g(x0)
        if abs(x1-x0) < delta:
            return x1,i
        x0 = x1
    print('method did not converge after 1000 iterations')


#PROGRAM ENDS HERE
```

```python

# Solve the following using either Newton-Raphson or Fixed-point

#PROGRAM STARTS HERE

import library

def invert_matrix_2x2(matrix):
    a, b = matrix[0][0], matrix[0][1]
    c, d = matrix[1][0], matrix[1][1]
    
    det = (a * d) - (b * c)
    if abs(det) < 1e-15:
        raise ValueError("Jacobian matrix is singular.")
        
    inv_det = 1.0 / det
    return [[ d * inv_det, -b * inv_det],[-c * inv_det,  a * inv_det]]

def newton_raphson2d(f1,f2,df1dx,df2dx,df1dy,df2dy,x0, y0, delta=1e-6):
    x, y = x0,y0
    
    for i in range(1000):
        f1val = f1(x,y)
        f2val = f2(x,y)
        f = [[f1val], [f2val]]

        df1dxval = df1dx(x,y)
        df1dyval = df1dy(x,y)
        df2dxval = df2dx(x,y)
        df2dyval = df2dy(x,y)

        jacobian = [[df1dxval,df1dyval],[df2dxval,df2dyval]]
        jacobian_inverse = invert_matrix_2x2(jacobian)

        step = library.matrix_multiply(jacobian_inverse, f)
        dx = step[0][0]
        dy = step[1][0]
        
        x = x - dx
        y = y - dy

        if (dx**2 + dy**2) ** 0.5 < delta:
            return x, y, i
            
    print("Method did not converge within the maximum iterations.")
    return None

def f1(x, y):
    return 3 * (x**2) - 2 * x * y - 3

def f2(x, y):
    return 3 * (y**2) - 4 * x * y

def df1dx(x, y):
    return 6 * x - 2 * y

def df1dy(x, y):
    return -2 * x

def df2dx(x, y):
    return -4 * y

def df2dy(x, y):
    return 6 * y - 4 * x

x_res, y_res, time = newton_raphson2d(f1,f2,df1dx,df2dx,df1dy,df2dy,2, 3)

print('zero of these functions is at ',x_res,y_res,' and number of iterations is ',time)


#PROGRAM ENDS HERE
```
