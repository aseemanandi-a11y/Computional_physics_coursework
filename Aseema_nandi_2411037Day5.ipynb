```python
# library function for bisection

#PROGRAM STARTS HERE

def bisection(f,a,b,delta = 1e-6):

    #checking unexpected cases
    if f(a)*f(b) > 0:
        print('no zero in bracket')
        return None
    if f(a)*f(b) == 0:
        if f(a) == 0:
            return a
        else:
            return b

    #actual method for 1000 max iterations
    for i in range(100):
        c = (a + b)/2
        if abs(f(c)) <= delta or (b - a)/2 <=0:
            return c
        if f(a)*f(c) < 0:
            b = c
        else:
            a = c
    print('zero not found within 100 iterations try different bracket values')
    return None

#PROGRAM ENDS HERE
```

```python
#library function for regula falsi

#PROGRAM STARTS HERE

def regfals(f,a,b,delta = 1e-6):

    #checking unexpected cases
    if f(a)*f(b) > 0:
        print('no zero in bracket')
        return None
    if f(a)*f(b) == 0:
        if f(a) == 0:
            return a
        else:
            return b

    c0 = a

    #actual method for 100 max iterations
    for i in range(100):
        c = b - ((b - a)*f(b)/(f(b) - f(a)))
        if abs(f(c)) <= delta or abs(c-c0)<=delta:
            return c
        c0 = c
        if f(a)*f(c) < 0:
            b = c
        else:
            a = c
    print('zero not found within 100 iterations try different bracket values')
    return None

#PROGRAM ENDS HERE
```

```python
#Find the root of the following function in the interval [1.5, 3.0] to an accuracy of 10−6 using both Bisection and Regula Falsi method.

#PROGRAM STARTS HERE

import library
import math

#function to be used f(x) = log(x/2) − sin(5x/2)
def f(x):
    return (math.log(x/2) - math.sin(5*x/2))

bisection_result = library.bisection(f,1.5,3.0)
regfal_result = library.regfals(f,1.5,3.0)

print("result from bisection method is", bisection_result)
print("result from regula falsi method is", regfal_result)


#PROGRAM ENDS HERE
```
